```python
from PyQt5.QtWidgets import *
from PyQt5.QtCore import *
import pandas as pd

import sys, os
import xlwings as xw

import numpy as np
import matplotlib as mpl
import matplotlib.pyplot as plt
# from scipy.interpolate import make_interp_spline
# from scipy import interpolate

plt.rcParams['axes.unicode_minus'] = False
plt.rcParams['font.family'] = 'Malgun Gothic'

from main_wnd import Ui_MainWindow
from mypandasModel import *

class Form(QMainWindow, Ui_MainWindow):
    def __init__(self):
        super().__init__()
        self.setupUi(self)

        idx = self.open_filter()
        print(idx)
        self.apply_lib(idx)

        self.btnGraph.clicked.connect(self.draw_graph)
        self.btnExcel.clicked.connect(self.make_excel)
                
        self.df = pd.DataFrame()
        self.lib = pd.DataFrame()
        self.inform = pd.DataFrame()
     

    def make_tableLib(self, df):
        model = pandasModel(df)
        self.tableLib.setModel(model)
        self.tableLib.resizeColumnsToContents()
        self.tableLib.horizontalHeader().setSectionResizeMode(QHeaderView.Stretch)
        self.tableLib.verticalHeader().setSectionResizeMode(QHeaderView.Stretch)       

    def make_tableConst(self, df):
        self.tableConst.setHidden(False)
        model = pandasModel(df)
        self.tableConst.setModel(model)
        self.tableConst.resizeColumnsToContents()
        self.tableConst.horizontalHeader().setSectionResizeMode(QHeaderView.Stretch)
        self.tableConst.verticalHeader().setSectionResizeMode(QHeaderView.Stretch) 

    def open_filter(self):
        PATH = "P:/개인자료/"
        file_name = "파이썬 활용 데이터 분석_신뢰성-IT.xlsx"

        book = xw.Book(PATH+file_name)
        df = book.sheets(4).used_range.options(pd.DataFrame).value

        self.df = pd.DataFrame(df.iloc[4:, :136].dropna(axis = 0, how = 'all').values)
        self.inform = self.df[[0, 1, 3, 5, 6, 8, 9]]
        self.inform.columns = ['기종', 'LOT', '구분', '모재', '조성', '온도', '전압']
        
        powder = list(set(self.inform["모재"]))
        powder = [x for x in powder if pd.isnull(x) == False]
        powder.sort()
        # filter_fir()
        txt = 'SHBT100'
        filtered = self.inform[self.inform["모재"] == txt]

        comp = list(set(filtered["조성"]))
        comp.sort()
        # filter_sec()
        txt = "NA392T010"
        filtered_sec = filtered[filtered["조성"] == txt]
        filtered_sec

        p_number = list(set(filtered_sec["기종"]))
        p_number.sort()
        # filter_thr()
        txt = "03A105MP3???"
        filtered_thr = filtered_sec[filtered_sec["기종"] == txt]
    
        lot = list(set(filtered_thr["LOT"]))
        lot.sort()
        # filter_four()
        txt = "AI17P56"
        filtered_four = filtered_thr[filtered_thr["LOT"] == txt]
        
        temp = list(set(filtered_four["온도"]))
        temp.sort()
        # filter_five()
        txt = 115.0
        filtered_five = filtered_four[filtered_four["온도"] == txt]

        volt = list(set(filtered_five["전압"]))
        volt.sort()
        # filter_six()
        txt = 16.0
        filtered_six = filtered_five[filtered_five["전압"] == txt]
        idx = filtered_six.index
        print('filtered_six', filtered_six, idx)

        filtered_six.reset_index(drop = True, inplace =True)
        filtered_show = filtered_six
        self.make_tableLib(filtered_show)
        # self.CaplineEdit.clear()
        return idx

    def apply_lib(self, idx):
        data = self.df.iloc[idx][data_cols].to_frame()
  

        # 보이는 테이블 전압 필터링
        # self.data_show = data
        self.make_tableLib(data)
       
    def draw_graph(self):
        x = [float(a) for a in self.fin['전압 (Vdc)'].values]
        y = self.fin['용량변화율 %'].values

        xs = np.linspace(x[0], x[-1], 1000)
        f = interpolate.interp1d(x, y, kind = 'linear')

        ys = f(xs)
        
        xp = [float(a) for a in self.low_show['전압 (Vdc)'].values]
        yp = self.low_show['용량변화율 %'].values
        
        plt.scatter(xp,yp, color = 'red')
        plt.plot(xs, ys)

        plt.xlabel('전압 (V)')
        plt.ylabel('용량변화율 %')

        plt.grid(True)
        plt.show()

    def make_excel(self):
        new_file = 'DC-bias library 분석'
        
        newfile = 'P:.\\..\\{}.xlsx'.format(new_file)
        # newfile = './{}.xlsx'.format(new_file)
        writer = pd.ExcelWriter(newfile)
        self.low.to_excel(writer, index=False, engine = 'openpyxl')
        writer.save()
        QMessageBox.information(self, '저장완료', '성공적으로 저장되었습니다.')
  
if __name__ == '__main__':
    app = QApplication(sys.argv)
    w = Form()
    w.show()
    app.exec_()
```
